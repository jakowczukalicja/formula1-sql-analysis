### relations between tables

```python
# Does every result have a driver record in a drivers table
query = """ 
        SELECT resultId
        FROM results
        WHERE driverId not in (SELECT driverId from drivers)
        """

pd.read_sql(query, conn)
```

```python
# How many drivers have more than one result record?
query = """ 
        SELECT COUNT(*) FROM (
            SELECT driverId
            FROM results
            GROUP BY driverId
            HAVING COUNT(*)>1
        )
        """

pd.read_sql(query, conn)
```

```python
# Does any driver have no result record?
query = """ 
        SELECT driverId 
        FROM drivers
        WHERE driverId not in (SELECT driverId FROM results)
        """

pd.read_sql(query, conn)
```

Thus, the drivers–results relationship is one-to-many:
- Each driver has one or more results (minimum: 1, maximum: many)
- Each result belongs to exactly one driver (minimum: 1, maximum: 1)

```python
# Does every result have a constructor record in a constructors table
query = """ 
        SELECT resultId
        FROM results
        WHERE constructorId not in (SELECT constructorId from results)
        """

pd.read_sql(query, conn)
```

```python
# Does any constructor have no result record?
query = """ 
        SELECT constructorId
        FROM constructors
        WHERE constructorId not in (SELECT constructorId FROM constructors)
        """

pd.read_sql(query, conn)
```

Thus, the constructors–results relationship is one-to-many:
- Each constructor has one or more results (minimum: 1, maximum: many)
- Each result belongs to exactly one constructor (minimum: 1, maximum: 1)
